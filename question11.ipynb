```python
def compute_mean(values):
    """
    Computes the mean of a list of numbers and prints the result.
    """
    mean_value = sum(values) / len(values)
    print(mean_value)

# Call the function
compute_mean([2, 4, 6, 8])
```

The code above calculates the average (mean) of a list of numbers. For example, `compute_mean([2, 4, 6, 8])` gives `5.0`.

The mean can be written in two equivalent ways:

\[
\bar{x} = \frac{x_1 + x_2 + \dots + x_n}{n} = \frac{1}{n} \sum_{i=1}^{n} x_i
\]

Here, \(x_1, x_2, \dots, x_n\) are the numbers in the list, and \(n\) is the total number of numbers.
